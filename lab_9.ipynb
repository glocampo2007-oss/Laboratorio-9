```python
plt.figure(figsize=(12, 6))
plt.plot(df['Mes'], df['Ventas'], marker='o', linestyle='-', color='skyblue')
plt.title('Ventas Mensuales (en miles)')
plt.xlabel('Mes')
plt.ylabel('Ventas (en miles)')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Actividad 2: Análisis de calificaciones de estudiantes

Primero, crearemos un DataFrame de pandas con los datos de calificaciones de los estudiantes proporcionados.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data_calificaciones = {"Estudiante": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Hannah", "Ian", "Jack"], "Matemáticas": [85, 92, 78, 90, 88, 76, 95, 89, 82, 94], "Ciencias": [90, 85, 80, 88, 92, 75, 95, 87, 81, 93], "Historia": [78, 82, 88, 90, 85, 80, 92, 84, 79, 91]}
df_calificaciones = pd.DataFrame(data_calificaciones)

display(df_calificaciones)
```

Ahora, utilizaremos las operaciones de NumPy para calcular algunas estadísticas clave de las calificaciones, como el promedio, el máximo y el mínimo para cada asignatura.

```python
promedio_matematicas = np.mean(df_calificaciones['Matemáticas'])
max_matematicas = np.max(df_calificaciones['Matemáticas'])
min_matematicas = np.min(df_calificaciones['Matemáticas'])

promedio_ciencias = np.mean(df_calificaciones['Ciencias'])
max_ciencias = np.max(df_calificaciones['Ciencias'])
min_ciencias = np.min(df_calificaciones['Ciencias'])

promedio_historia = np.mean(df_calificaciones['Historia'])
max_historia = np.max(df_calificaciones['Historia'])
min_historia = np.min(df_calificaciones['Historia'])

print(f"--- Estadísticas de Matemáticas ---")
print(f"Promedio: {promedio_matematicas:.2f}")
print(f"Máxima: {max_matematicas}")
print(f"Mínima: {min_matematicas}")
print(f"\n--- Estadísticas de Ciencias ---")
print(f"Promedio: {promedio_ciencias:.2f}")
print(f"Máxima: {max_ciencias}")
print(f"Mínima: {min_ciencias}")
print(f"\n--- Estadísticas de Historia ---")
print(f"Promedio: {promedio_historia:.2f}")
print(f"Máxima: {max_historia}")
print(f"Mínima: {min_historia}")
```

Finalmente, visualizaremos las calificaciones promedio por asignatura para identificar el rendimiento general.

```python
promedios_asignaturas = df_calificaciones[['Matemáticas', 'Ciencias', 'Historia']].mean()

plt.figure(figsize=(10, 6))
promedios_asignaturas.plot(kind='bar', color=['lightcoral', 'lightgreen', 'cornflowerblue'])
plt.title('Promedio de Calificaciones por Asignatura')
plt.xlabel('Asignatura')
plt.ylabel('Calificación Promedio')
plt.ylim(0, 100) # Establecer el límite del eje y de 0 a 100 para calificaciones
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--')
plt.tight_layout()
plt.show()
```
